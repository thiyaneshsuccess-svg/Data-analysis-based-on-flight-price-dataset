This is the trickiest cleaning step because the column mixes three formats. We write a
small function and apply it to every row with `.apply()` — this is the standard pandas
pattern for "no built-in function does exactly this, so I'll write my own logic."

We use Python's `re` (regular expressions) module to detect and pull apart each format.

```python
import re
import numpy as np

def parse_duration_to_minutes(value):
    """
    Convert a Duration string into total minutes, handling three formats:
      - "3h 11m"   -> hours and minutes explicitly labeled
      - "177 min"  -> already in minutes
      - "3.34"     -> decimal hours (e.g. 3.34 hours)
    Returns np.nan if the value is missing or doesn't match any known format.
    """
    if pd.isna(value):
        return np.nan

    value = str(value).strip()

    # Format 1: "Xh Ym" (Y is optional, e.g. "16h" alone can happen too)
    hm_match = re.match(r'^(\d+)h\s*(\d*)m?$', value)
    if hm_match:
        hours = int(hm_match.group(1))
        minutes = int(hm_match.group(2)) if hm_match.group(2) else 0
        return hours * 60 + minutes

    # Format 2: "Y min"
    min_match = re.match(r'^(\d+)\s*min$', value)
    if min_match:
        return int(min_match.group(1))

    # Format 3: plain decimal number -> treat as decimal hours
    decimal_match = re.match(r'^\d+\.?\d*$', value)
    if decimal_match:
        return float(value) * 60

    # Anything else we don't recognize
    return np.nan

# .apply() runs this function once per row in the column and returns a new Series
df_clean['Duration_Minutes'] = df_clean['Duration'].apply(parse_duration_to_minutes)

# Check a sample to confirm the conversion looks right across all 3 original formats
df_clean[['Duration', 'Duration_Minutes']].sample(15, random_state=1)
```

```python
df_clean = df.copy()

# --- Drop duplicate rows ---
before = len(df_clean)
df_clean = df_clean.drop_duplicates()
print(f"Dropped {before - len(df_clean)} duplicate rows. New shape: {df_clean.shape}")

# --- Standardize Airline casing ---
df_clean['Airline'] = df_clean['Airline'].str.strip().str.title()
df_clean['Airline'] = df_clean['Airline'].replace({'Airasia India': 'AirAsia India'})

# --- Standardize Source/Destination (city name vs airport code vs "X Airport") ---
city_map = {
    'AMD': 'Ahmedabad', 'Ahmedabad Airport': 'Ahmedabad',
    'BLR': 'Bangalore', 'Bangalore Airport': 'Bangalore',
    'BKK': 'Bangkok', 'Bangkok Airport': 'Bangkok',
    'BOM': 'Mumbai', 'Mumbai Airport': 'Mumbai',
    'CCU': 'Kolkata', 'Kolkata Airport': 'Kolkata',
    'DEL': 'Delhi', 'Delhi Airport': 'Delhi',
    'DOH': 'Doha', 'Doha Airport': 'Doha',
    'DXB': 'Dubai', 'Dubai Airport': 'Dubai',
    'FRA': 'Frankfurt', 'Frankfurt Airport': 'Frankfurt',
    'GOI': 'Goa', 'Goa Airport': 'Goa',
    'HYD': 'Hyderabad', 'Hyderabad Airport': 'Hyderabad',
    'JAI': 'Jaipur', 'Jaipur Airport': 'Jaipur',
    'JFK': 'New York', 'New York Airport': 'New York',
    'LHR': 'London', 'London Airport': 'London',
    'MAA': 'Chennai', 'Chennai Airport': 'Chennai',
    'PNQ': 'Pune', 'Pune Airport': 'Pune',
    'SIN': 'Singapore', 'Singapore Airport': 'Singapore',
    'SYD': 'Sydney', 'Sydney Airport': 'Sydney',
}
df_clean['Source'] = df_clean['Source'].replace(city_map)
df_clean['Destination'] = df_clean['Destination'].replace(city_map)

# --- Standardize Total_Stops into a clean integer ---
stops_map = {'non-stop': '0', '1 stop': '1', '2 stops': '2'}
df_clean['Total_Stops'] = df_clean['Total_Stops'].replace(stops_map)
df_clean['Total_Stops'] = pd.to_numeric(df_clean['Total_Stops'], errors='coerce')

# --- Clean Price (strip "Rs." prefix and commas) ---
df_clean['Price'] = df_clean['Price'].astype(str).str.replace(r'[^0-9.]', '', regex=True)
df_clean['Price'] = pd.to_numeric(df_clean['Price'], errors='coerce')

# --- Clean Passenger_Count (word numerals -> digits) ---
word_to_num = {'one': '1', 'two': '2', 'three': '3', 'four': '4', 'five': '5', 'six': '6'}
df_clean['Passenger_Count'] = df_clean['Passenger_Count'].replace(word_to_num)
df_clean['Passenger_Count'] = pd.to_numeric(df_clean['Passenger_Count'], errors='coerce')

# --- Convert remaining numeric-looking columns ---
df_clean['Distance_km'] = pd.to_numeric(df_clean['Distance_km'], errors='coerce')
df_clean['Days_Before_Departure'] = pd.to_numeric(df_clean['Days_Before_Departure'], errors='coerce')

# --- Parse Departure_Date into a real datetime ---
df_clean['Departure_Date'] = pd.to_datetime(df_clean['Departure_Date'], errors='coerce')

print(df_clean.dtypes)
```
