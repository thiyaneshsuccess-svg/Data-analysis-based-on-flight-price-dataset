```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
df=pd.read_csv('flight_pricing_dataset.csv')
df.head()
```

```python
missing_counts = df.isnull().sum().sort_values(ascending=False)
print(missing_counts)
print("\nMissing %:")
print((missing_counts / len(df) * 100).round(2))
```

```python
print("Fully duplicated rows:", df.duplicated().sum())
```

```python
print("Duplicate Flight_IDs:", df['Flight_ID'].duplicated().sum())
```

**What we found so far:**
- Roughly 5% of values are missing in almost every column (not concentrated in one place)
- ~2,000 fully duplicated rows
- Every column is stored as text (`object`), even numeric ones like `Price` and `Distance_km`
